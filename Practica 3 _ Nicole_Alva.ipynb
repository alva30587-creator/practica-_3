```python
from typing_extensions import Counter
import re
##Pregunta 1

texto = """
Juan: 987654321
María: 912345678
Pedro: 998877665
Ana: 923456789
"""
numero= re.findall(r'9\d{8}', texto)
print(numero)
print("El total de numeros es: ",len(numero))

```

```python
## Pregunta 2
texto = """
correo1@gmail.com
ventas@empresa.pe
contacto@universidad.edu.pe
soporte@hotmail.com
"""
correo = re.findall(r'[\w\.]+@[\w]+\.[a-zA-Z]{2,}',texto)
print(correo)
print("El total de correos encontrados fueron :",len(correo))
```

```python
## Pregunta 3
import requests
from bs4 import BeautifulSoup
import pandas as pd
enlaces= "https://books.toscrape.com"
respu= requests.get(enlaces)
respu.encoding = "utf-8"
soup = BeautifulSoup(respu.text, "html.parser")

print(soup.title)
print(soup.title.text)

#Obtenga información de los primeros 20 libros mostrados en la página principal.
## Titulos
titulos = [a.get("title") for a in soup.select("h3 a")]
print(titulos)

## Precios
precios = [p.get_text(strip=True) for p in soup.find_all("p", class_="price_color")]
print(precios)

## Pregunta 4
datos = []
for i in range(len(titulos)):
    datos.append({
        "Titulo": titulos[i],
        "Precio": precios[i]
    })

df = pd.DataFrame(datos)
print("\nDataframe:")
print(df)

## Se imprime en formato CSV
df.to_csv("libros.csv", index=False, encoding="utf-8-sig")
```

```python
## Pregunta 5

precios_limp = [float(re.findall(r"\d+\.\d+", p)[0]) for p in precios]

print(precios_limp)
menor = min(precios_limp)
mayor = max(precios_limp)
print("El precio menor es: ",menor)
print("El precio mayor es: ",mayor)
```
